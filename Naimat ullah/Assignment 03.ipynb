1. Write a program that accepts a list from user and print the alternate element of list.

```python
a=input("enter a list: ").split()
print("The list is:", a)
print(a[::2])

#we can solve same problem using for loop
b=[]
for i in range(0, len(a), 2):
    b.append(a[i])
print(b)
```

2. Write a program that accepts a list from user. Your program should reverse the content of list and 
display it. Do not use reverse() method. 

```python
c=input("enter a list: ").split()
print("The list is:", c)
d=[]
for i in range(len(c)-1,-1,-1):
    d.append(c[i])
print(d)
```

3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard. 

```python
e=input("enter a list: ").split()
e=[int(i) for i in e]
print("The list is:", e)
f=e[0]
for i in e:
    if i>f:
        f=i
print("The largest element in the list is:", f)
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.  

```python
g=[10,20,30,40,50]
h=len(g)
j=[]
for i in range(h):
    if i<h-1:
        j.insert(i+1,g[i])
    else:
        j.insert(0,g[i])
print("The new list is:", j)
```

9. Write a program to add two matrices of size n x m. 

```python
a=int(input("enter a number a number of rows: "))
b=int(input("enter a number of columns: "))
matrixA=[]
matrixB=[]
matrixC=[]
for i in range(a):
    row=[]
    for j in range(b):
        row.append(int(input("enter a number: ")))
    matrixA.append(row)
print("The first matrix is:")
print(matrixA)
for i in range(a):
    row=[]
    for j in range(b):
        row.append(int(input("enter a number: ")))
    matrixB.append(row)
print("The second matrix is:")
print(matrixB)
for i in range(a):
    row=[]
    for j in range(b):
        row.append(matrixA[i][j]+matrixB[i][j])
    matrixC.append(row)
print("The sum of the two matrices is:")
print(matrixC)
```

10. Write a program to multiply two matrices 

```python
c=[[2,5],
   [3,4]]
d=[[5,3],
   [9,10]]
e=[[0,0],
   [0,0]]
for i in range(2):
    for j in range(2):
        for k in range(2):
            e[i][j] = e[i][j] + c[i][k] * d[k][j]
print("first matrix is:", c)
print("second matrix is:", d)
print("The product of the two matrices is:")
print(e)
```
