Q5 — Word Counter
Take sentence input and count word frequency.


```python
sentence = input (" enter any sentence :")

word = sentence . split()

count={}

for words in word:
    if words in count :
        count[words] +=1
    else:
        count[words] =1 

print(count)            
```

6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.

```python
marks =  {
    "Math":80,
    "English":70,
    "Physics":90
    }

total = sum(marks . values())

average = total / len(marks)

print ("TOTAL", total)

print("AVERAGE :", average)
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student.

```python
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

highest = 0 
name = ""

for student in students:
    if students[student] > highest :
        highest = students[student]
        name = student 




print("highest scorring student :",name)
print("score: ",highest)        
```
