```python
# [] -- lists (ordered by index)      {} -- dictionaries (key: value pairs)
# a dict groups related data under one variable, accessed by key instead of index
user = {
    "name": "Sarah",
    "plan": "Pro",
    "active": True,
    "storage_gb": 250,
    "marks": [10,20,60,80]
}
```

```python
# access values by key, not by index
print(user["name"])
print(user["marks"])
print(user["active"])
```

```python
# "age" isn't a key in user -- direct [] access raises KeyError for a missing key
print(user["age"])
```

### get function

```python
# get(key, default) -- returns default instead of raising KeyError when the key is missing
print(user.get("age", "N/A"))
print(user.get("marks", "N/A"))
```

```python
user = {
    "name": "Sarah",
    "plan": "Pro",
    "active": True,
    "storage_gb": 250,
    "marks": [10,20,60,80]
}

user["name"] = "Zafar"  # dicts are mutable -- assigning to an existing key updates its value
print(user)
```

```python
user["active"] = False  # same pattern: dict[key] = new_value
print(user)
```

```python
# pop(key) -- removes the key and returns its value
removed = user.pop('marks')
print(removed)
```

```python
# 'marks' key is gone after pop()
print(user)
```

```python
# del -- removes a key without returning its value (unlike pop)
del user["storage_gb"]

print(user)
```

### Dictionary Methods

```python
user = {
    "name": "Sarah",
    "plan": "Pro",
    "active": True,
    "storage_gb": 250,
    "marks": [10,20,60,80]
}
```

```python
# keys() -- returns a view of all keys in the dict

user.keys()
```

```python
# values() -- returns a view of all values in the dict

user.values()
```

```python
# items() -- returns a view of (key, value) pairs, useful for looping over both together
user.items()
```
